```python
# Simulador de IA Financiera - Análisis de CAT (Costo Anual Total)
# Nivel: Principiante (Solo while, if, elif, else)

print("====================================================")
print("¡Bienvenido al Asistente Virtual de IA Financiera!")
print("Analizador del Costo Anual Total (CAT)")
print("====================================================")
print("Este programa te ayudará a saber si una tarjeta o préstamo te conviene.")

ejecutando = True

# Ciclo principal para mantener la app interactiva
while ejecutando:
    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Analizar un nuevo producto financiero (CAT)")
    print("2. Ver consejos generales sobre el CAT")
    print("3. Salir del asistente")

    opcion = input("Selecciona una opción (1-3): ")

    if opcion == "1":
        print("\n--- NUEVO ANÁLISIS ---")
        nombre_producto = input("¿Qué producto vas a analizar? (Ej. Tarjeta Oro, Préstamo Familiar): ")

        # Pedimos los datos numéricos en porcentaje (escribir solo el número, ej: 45)
        tasa_interes = float(input("Ingresa la tasa de interés anual (%): "))
        comisiones = float(input("Ingresa el costo de comisiones o anualidad anual (%): "))

        # Cálculo simplificado del CAT (Suma de tasa + costos asociados)
        cat = tasa_interes + comisiones

        print("\n----------------------------------------------------")
        print("Resultado: El CAT estimado para '" + nombre_producto + "' es de: " + str(cat) + "%")
        print("----------------------------------------------------")

        # Aquí entra la "IA" a analizar los datos con estructuras if-elif-else
        print("[IA Financiera procesando diagnóstico...]")

        if cat <= 20:
            print("Diagnóstico: ¡Excelente! Este CAT es muy bajo y amigable.")
            print("Consejo de IA: Es una gran opción de financiamiento. Aprovéchala, pero paga a tiempo.")

        elif cat > 20 and cat <= 50:
            print("Diagnóstico: CAT Moderado/Aceptable.")
            print("Consejo de IA: Está dentro del promedio del mercado. Úsala con responsabilidad y evita acumular deudas.")

        elif cat > 50 and cat <= 100:
            print("Diagnóstico: ¡ALERTA! CAT Muy Alto.")
            print("Consejo de IA: Este producto ya es caro. Si compras algo a meses, asegúrate de que no te genere más intereses o terminarás pagando casi el doble.")

        else:
            print("Diagnóstico: ¡PELIGRO FINANCIERO! CAT Extremo.")
            print("Consejo de IA: ¡Huye de aquí! Este interés es un riesgo enorme para tus bolsillos. Busca otra alternativa urgentemente.")

    elif opcion == "2":
        print("\n--- CONSEJOS DE LA IA SOBRE EL CAT ---")
        print("* El CAT incluye TODO: tasa de interés, anualidad, seguros y comisiones.")
        print("* Siempre compara el CAT de dos tarjetas antes de firmar; elije siempre el menor.")
        print("* Si eres 'totalero' (pagas todo lo que gastas al mes), el CAT no te afectará tanto, pero igual cuida las comisiones fijas.")

    elif opcion == "3":
        print("\nGracias por usar el Asistente de IA Financiera. ¡Cuida tu dinero! ¡Adiós!")
        ejecutando = False # Rompe el ciclo while para cerrar el programa

    else:
        print("\n[Opción Incorrecta] Por favor, selecciona un número del 1 al 3.")
```

```python
print("====================================================")
print("¡Bienvenido al Asistente Virtual de IA Financiera!")
print("Analizador del Costo Anual Total (CAT)")
print("====================================================")
print("Este programa te ayudará a saber si una tarjeta o préstamo te conviene.")

# Variable para mantener el programa activo
ejecutando = True

# Ciclo principal
while ejecutando:

    print("\n--- MENÚ PRINCIPAL ---")
    print("1. Analizar un nuevo producto financiero (CAT)")
    print("2. Ver consejos generales sobre el CAT")
    print("3. Salir del asistente")

    opcion = input("Selecciona una opción (1-3): ")

    # OPCIÓN 1
    if opcion == "1":

        print("\n--- NUEVO ANÁLISIS ---")

        nombre_producto = input(
            "¿Qué producto vas a analizar?: "
        )

        tasa_interes = float(
            input("Ingresa la tasa de interés anual (%): ")
        )

        comisiones = float(
            input("Ingresa el costo de comisiones (%): ")
        )

        # Cálculo del CAT
        cat = tasa_interes + comisiones

        print("\n----------------------------------------------------")
        print("Resultado: El CAT estimado para",
              nombre_producto,
              "es de:",
              cat, "%")
        print("----------------------------------------------------")

        print("\n[IA Financiera procesando diagnóstico...]")

        # Diagnóstico
        if cat <= 20:
            print("Diagnóstico: ¡Excelente!")
            print("Consejo: Muy buena opción financiera.")

        elif cat <= 50:
            print("Diagnóstico: CAT Moderado.")
            print("Consejo: Úsala con responsabilidad.")

        elif cat <= 100:
            print("Diagnóstico: ¡ALERTA!")
            print("Consejo: Puede salir muy caro.")

        else:
            print("Diagnóstico: ¡PELIGRO FINANCIERO!")
            print("Consejo: Busca otra alternativa.")

    # OPCIÓN 2
    elif opcion == "2":

        print("\n--- CONSEJOS SOBRE EL CAT ---")

        print("* El CAT incluye intereses y comisiones.")
        print("* Siempre compara diferentes opciones.")
        print("* Entre menor CAT, mejor para ti.")

    # OPCIÓN 3
    elif opcion == "3":

        print("\nGracias por usar el asistente financiero.")
        print("¡Adiós! 👋")

        # Termina el ciclo
        ejecutando = False

    # OPCIÓN INVÁLIDA
    else:

        print("\n[ERROR] Opción incorrecta.")
        print("Selecciona un número del 1 al 3.")
```
